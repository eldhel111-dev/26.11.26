## 선형회귀 실습

### 학습 데이터 준비

```python
import pandas as pd
import matplotlib.pyplot as plt


#선형회귀 모델 임포트
from sklearn.linear_model import LinearRegression

# 한글 폰트 설정
plt.rcParams["font.family"] = "Malgun Gothic"
# plt.rcParams["font.family"] = "AppleGothic"


# 데이터
data = {
    "hours": [1, 2, 3, 4, 5],
    "scores": [47, 53, 65, 73, 87]
}

# 판다스 데이터프레임으로 전환
df = pd.DataFrame(data)

df
```

### 입력 데이터와 정답데이터 분리



```python
# 입력 데이터 (Feature) : 2차원 - 독립변수 (왜 독립변수?)
X = df[["hours"]]   # 공부시간

# 출력 데이터 (Target) : 1차원 종속변수
y = df["scores"]    # 시험점수
```

### 모델 생성 및 학습

```python
# 선형회귀 모델 객체 생성
model = LinearRegression()

# 모델 학습
model.fit(X, y)
```

### 결과 확인

```python
# 다중 선형 회귀일 경우 결과값이 배열로 리턴 뭐가 결과값이 배열로 나오는지? y= wx + b ?
# 기울기(Weight) : 배열 리턴
print(f"기울기: {model.coef_[0]:.2f}") # < [0] 해야 2f 되는이유 : 리스트에서 꺼내오는거라?

# 절편(Bias)
print(f"절편: {model.intercept_:.2f}")
```

### 새로운 데이터로 예측하기

```python
# 6시간 학습한 학생의 성적 예측
# 독립 변수 X(6)추가
new_data = pd.DataFrame({"hours": [6]})

predicted_score = model.predict(new_data)
print(f"6시간 학습한 학생의 예상 성적: {predicted_score[0]:.2f}")
```

### 학습 결과 시각화

```python
# 학습 데이터 대한 예측값

y_pred = model.predict(X)
print("학습 데이터에 대한 예측값:", y_pred)

plt.Figure(figsize=(8,5))

# 실제 성적 (산점도)
plt.scatter(X["hours"], y, color="blue", label="실제 성적")

# 예측한 성적 (회귀직선)
plt.plot(X["hours"], y_pred, color="red", label="예측 성적")

# 6시간 학습한 학생의 예측성적
plt.scatter(6, predicted_score[0], color="green", label="6시간 성적")

plt.legend()
plt.show()
```

## 모델 평가


```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

# MAE : 오차 절대 평균
# 오차의 크기를 직관적으로 이해
mae = mean_absolute_error(y, y_pred)
print(f"학습 데이터에 대한 평균 절대 오차(MAE)){mae:.2f}")

# MSE : 오차 제곱 평균
# 제곱하는 이유는 큰 오차에 대해서 민감하게 반응
mse = mean_squared_error(y, y_pred)
print(f"학습 데이터에 대한 평균 제곱 오차(MAE)){mse:.2f}")

# R2 Score : 결정계수
# 모델이 실제 데이터를 얼마만큼 정확한지를 나타내는 지표
# 0 <==> 1
# 음수일 경우 학습 실패
r2 = r2_score(y, y_pred)
print(f"학습 데이터에 대한 결정 계수(R^2)){r2:.2f}")
```

### 손실 함수 (Loss Function)

- 모델이 예측한 값이 실제 정답과 얼마나 차이가 나는지를 수치로 표현하는 함수
- MSE 를 대표적으로 사용









### 학습 데이터 준비


```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
# 배치 하강법 (BGD) : 전체 데이터를 사용, 시간, 리소스 많이 사용
# 확률적 경사하강법 (SGD) : 확률적을 1개의 데이터를 추출해서 하강 방법
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import mean_squared_error

# 학습 데이터 생성
df = pd.DataFrame({
    "hours": [1, 2, 3, 4, 5],
    "scores": [47, 53, 65, 73, 87]
})

# 독립변수 (입력변수)
X = df[["hours"]]
# 종속변수 (정답)
y = df["scores"]

df
```

```python
# 선형 회귀
model = LinearRegression()
# 모델학습
model.fit(X, y)
# 예측값
y_pred = model.predict(X)
# 결과값
print(f"Weight: {model.coef_[0]:.2f}")
print(f"Bias: {model.intercept_:.2f}")
print(f"MSE {mean_squared_error(y, y_pred):.4f}")
```

### 확률적 경사하강법 모델 생성

- SGDRegressor 모델생성
- SGD(Stochastic Gradient Descent)


```python
sgd = SGDRegressor(
    # 제곱 오차 기반 손실함수
    loss="squared_error",

    # 학습률을 일정하게 유지
    learning_rate="constant",

    # 초기 학습률
    eta0=0.01,

    # 학습횟수 fit() 반복 회수
    max_iter=1000,

    # 학습 데이터 섞지않음
    shuffle=False,

    # 난수 시드 설정
    random_state=42
)
```

```python
# 전체 데이터를 기준으로 총 몇회 반복

epochs = 1000


# 학습과정 저장 리스트
history = []

for epochs in range(epochs):
    # 전체 데이터 1회 학습 (1 epoch)
    sgd.partial_fit(X, y)

    # 1회 학습 후 예측
    y_pred = sgd.predict(X)

    # 현재 모델의 MSE 계산
    mse = mean_squared_error(y, y_pred)

    # 현재 학습 결과 저장(누적)
    history.append({
        "epoch": epochs + 1,
        "weight": sgd.coef_[0],
        "bias": sgd.intercept_[0],
        "mse": mse
    })

# 학습 기록 => DataFrame 으로 변환
history_df = pd.DataFrame(history)

history_df
```

```python
history_df[history_df["epoch"].isin([1, 5, 10 , 50, 100, 300, 500, 1000])]
```

### 손실함수 차트



```python
plt.figure(figsize=(10,5))

# Epoch 따른 MSE 변화
plt.plot(
    history_df["epoch"],
    history_df["mse"],
    color="blue",
    label="학습손실(Traing Loss)"    
)

plt.xlabel("Loss")
plt.ylabel("MSE")
plt.grid(alpha=0.3)
plt.show()
```

### 기울기화 절편 변화 시각화

```python
fig, axes = plt.subplots(1, 2, figsize= (12,5))

# 기울기 변화
axes[0].plot(
    history_df["epoch"],
    history_df["weight"],
    color="blue",

)
axes[0].grid(alpha=0.3)                     # grid() 란?
axes[0].set_ylabel("weight")

# 절편 변화
axes[1].plot(
    history_df["epoch"],
    history_df["bias"],
    color="blue"
)
axes[1].grid(alpha=0.3)
axes[1].set_ylabel("Bias")

plt.show()
```

### 최종 모델로 시험 적적 예측하기  

```python
# 최종 학습된 기울기, 절편
print(f"기울기(w): {sgd.coef_[0]:.2f}")
print(f"절편(b): {sgd.intercept_[0]:.2f}")

# 6시간 학습한 학생
new_data = pd.DataFrame({"hours": [6]})

# 경사하강법으로 학습한 모델로 예측
prediction = sgd.predict(new_data)

print(f"예측 성적: {prediction[0]:.2f}")
```

## 다중 선형회귀 [Multiple linear Regression]

### 라이브러리 임포트 및 데이터 준비

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

# 데이터 생성
df = pd.DataFrame({    
    "study_hours":    [1, 2, 3, 4, 5, 2, 3, 5, 4, 1],
    "sleep_hours":    [6, 8, 5, 7, 8, 5, 8, 6, 6, 8],
    "practice_count": [10, 15, 20, 25, 30, 12, 28, 22, 32, 8],
    "score":          [47, 58, 60, 76, 87, 49, 73, 77, 79, 52]
})

df
```

## 상관관계 확인 (차트)

```python
# 1행 3열의 그래프 영역 생성
fig, axes = plt.subplots(1, 3, figsize=(10, 5))
    
# 공부시간 + 성적
sns.scatterplot(
    data = df,
    x= "study_hours",
    y= "score",
    ax=axes[0]
)


# 수면시간 + 성적
sns.scatterplot(
    data = df,
    x= "sleep_hours",
    y= "score",
    ax=axes[1]
)


# 문제풀이 개수 + 상작
sns.scatterplot(
    data = df,
    x= "practice_count",
    y= "score",
    ax=axes[2]
)

plt.show()
```
