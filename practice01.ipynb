```python
# 미션 4. 이상치 계산 및 데이터 필터링

# 목표
# Titanic 승선요금(fare)의 이상치를 계산하고
# 이상치를 제외한 새로운 데이터프레임 만들기

# 요구사항
# 1. fare 컬럼의 Q1(25%) 계산
# 2. fare 컬럼의 Q3(75%) 계산
# 3. IQR = Q3 - Q1 계산
# 4. 이상치 상한선 = Q3 + 1.5 * IQR 계산
# 5. 상한선을 초과하는 행의 개수 출력
# 6. 상한선을 초과하는 행을 제외하고 copy()로 복사
# 7. 필터링 전후 행 개수 출력

# 사용 문법
# quantile(), len(), copy(), 조건식

# 원본 데이터 초기화
import seaborn as sns
df = sns.load_dataset("titanic")

# 아래부터 직접 작성
print(df["fare"].quantile([0.25, 0.75]))


q1 = df["fare"].quantile(0.25)
q3 = df["fare"].quantile(0.75)
iqr = q3 - q1

print(f"{q1},{q3},{iqr}")
```
